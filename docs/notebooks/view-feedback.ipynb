## Suggested trial

1. **Interactive figure:** move `I`, click the plot, then try Inspect and Expand. Close the workspace; the changes should still be there.
2. **16:9 experiment bench:** choose a tool in the inspector. Can you watch both plots while changing settings?
3. **Instrument drawer:** open Controls, adjust `a`, then close it. Is the extra space worth the changing plot width?
4. **Comparison lens:** Freeze scene, then change `I`. The muted nullclines and dashed paths retain the old experiment. Try Show frozen scene.
5. **Guided lesson:** switch between Explore, Observe and Compare. Optionally run a small sweep (4–8 samples). Parameters, frozen reference and completed results survive layout switches; leaving a running sweep cancels it.

6. **Small multiples:** edit three parameter values, then Use a tile to apply it. Editing alone should not change the live parameter.
7. **Bifurcation navigator:** run a 4–8 sample scan, then select values by clicking or keyboard. This is sampled exploration, not continuation.
8. **Initial-condition explorer:** click several starts and compare their numbered paths. This is not a classified basin map.
9. **Evidence notebook:** capture a labeled observation, change settings, preview it, then Restore. Download JSON before Reset if you want to keep it.
10. **Prediction & reveal:** write a prediction and reveal the intentionally hidden trajectory. Changing settings should hide it again.
11. **Stochastic ensemble:** compare seeded traces, change noise strength, and explicitly Use a seed. Traces are not confidence intervals.

Try resizing the notebook window. Wide workspaces use 16:9; narrow ones stack. Reset restores the first frontend mount's initial experiment, not your last selected view. Rerunning the creation cell creates a new experiment.
